# 00 Setup check
Imports every installed package and prints its version.

```python
import sys, importlib, importlib.metadata as md
print("Python", sys.version.split()[0])
pkgs = {"pandas":"pandas","numpy":"numpy","networkx":"networkx","scikit-learn":"sklearn","xgboost":"xgboost",
 "lightgbm":"lightgbm","matplotlib":"matplotlib","plotly":"plotly","folium":"folium","streamlit":"streamlit",
 "shap":"shap","jupyter":"jupyter","requests":"requests","beautifulsoup4":"bs4","pyarrow":"pyarrow",
 "torch":"torch","torch_geometric":"torch_geometric"}
for dist, mod in pkgs.items():
    try:
        importlib.import_module(mod)
        print(f"OK   {dist:16s} {md.version(dist)}")
    except Exception as e:
        print(f"FAIL {dist:16s} {type(e).__name__}: {e}")
```
